```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_excel("./resources/Charting.xlsx")
data.info()
```

```python
data["Date"] = pd.to_datetime(data["Date"])
data.info()
```

```python
data["ReturnsOverNight"] = data["Open"].shift(-1) / data["S&P 500 - Close"] - 1
data["ReturnsDay"] = data["S&P 500 - Close"] / data["Open"] - 1

df = data

df["A_ReturnsOverNight"] = df["ReturnsOverNight"].cumsum()
df["A_ReturnsDay"] = df["ReturnsDay"].cumsum()

df.drop(df.index[-1], inplace=True)
df.tail(2)
```
