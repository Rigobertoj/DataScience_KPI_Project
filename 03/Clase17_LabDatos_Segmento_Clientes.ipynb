Este enfoque, aunque fácil de implementar, solo considera una variable: el dinero que gastan los clientes. Esto tiene varios problemas:

- **Simplicidad excesiva**: Estamos ignorando otras variables clave que podrían ser importantes, como:

    - **Recency (Recencia)**: ¿Cuándo fue la última compra de un cliente? Un cliente que gastó mucho dinero hace un año podría ser menos relevante que un cliente que gastó una cantidad moderada recientemente.

    - **Frecuencia**: ¿Con qué frecuencia compran los clientes? Un cliente que compra regularmente podría ser más valioso que alguien que hace compras grandes pero raras.

    - **Cantidad de órdenes**: Los clientes que realizan muchas compras pequeñas podrían ser más valiosos que aquellos que hacen una sola compra grande.

- **Limitación en el análisis**: Este enfoque solo utiliza una dimensión de los datos (el dinero gastado), por lo que no proporciona una imagen completa del comportamiento de los clientes.

##### **3. Segmentación por Clustering**

El **clustering** automatiza el proceso de segmentación considerando todas las variables de interés al mismo tiempo. En lugar de hacer divisiones arbitrarias como los cuartiles, el algoritmo agrupa a los clientes en grupos similares.

**¿Qué es clustering?**

Clustering es una técnica de aprendizaje no supervisado que **agrupa datos** en conjuntos (o clústeres) que son **similares entre sí**.

Es decir, los elementos dentro de un clúster son **más parecidos entre sí que con los elementos de otros clústeres.**

    <img src="https://i.stack.imgur.com/cIDB3.png"  width="400" />

**Conceptos clave:**

- Cluster: Un grupo de puntos similares entre sí.
- Centroide: El punto central de un cluster.
- Distancia Euclidiana: Una forma de medir qué tan cerca están dos puntos.
- Número de Clusters (k): Cantidad de grupos a formar.
- Varianza Intra-cluster: Qué tan juntos están los puntos dentro de un cluster.

**AHORA, ENTENDAMOS COMO FUNCIONA UN CLUSTER - KMEANS**

**Pre-procesamiento de los datos**
- Revisar si existe o no aoutliers o valores extremos por la sensibilidad de KMeans a estos.
- Revisar si es necesario normalizar los datos.
- Revisar si es necesario reducir la dimensionalidad de los datos.
- Tratamiento de datos nulo o faltante.

**Algoritmo KMeans**
- **Paso 1:** Seleccionar el número de clusters (k).
- **Paso 2:** Inicializar los centroides de los clusters.
- **Paso 3:** Calcular la distancia de cada punto al centroide.
- **Paso 4:** Asignar cada punto al centroide más cercano.
- **Paso 5:** Recalcular los centroides.
- **Paso 6:** Repetir los pasos 3, 4 y 5 hasta que los centroides no cambien.

**Revisiones finales**
- Revisar la varianza intra-cluster.
- Revisar la cohesión y separación de los clusters.
- Revisar si los clusters son interpretables.

##### **4. Clustering en practica - KMeans**

La empresa **SmartElectro**, especializada en la venta de aparatos electrónicos y artículos coleccionables, busca hacer sus campañas de marketing más efectivas, seleccionando de forma más precisa a su público objetivo.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

clientes = pd.read_excel('./resources/Clientes_Clustering.xlsx')
clientes.head()
```

```python
filtered = clientes[['Ordenes','AVG_Orden_USD','Recencia']]
filtered.head()
```

```python
filtered.columns
```

```python
#pip install scikit-learn
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaled_df = pd.DataFrame(
    scaler.fit_transform(filtered),
    columns=filtered.columns
)

scaled_df.head()
```

```python
scaled_df.describe().round(2)
```

**Número de grupos**

- **Método del Codo:** Calcula la suma de los errores cuadráticos (SSE) para diferentes valores de k.
- **Método de la Silueta:** Evalúa qué tan bien están agrupados los puntos dentro de sus clusters.
- **Conocimiento del Dominio:** Conocer el contexto del problema puede ayudar a elegir k. Por ejemplo, si se analizan tipos de clientes, es posible que se espere un número aproximado de grupos.

```python
from sklearn.cluster import KMeans

fig, [ax1, ax2] = plt.subplots(1, 2, figsize=(12, 6))

inertia = []

k_range = range(1, 30)

for k in k_range:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(scaled_df)
    inertia.append(model.inertia_)

inertia_ratio = np.diff(inertia) / np.array(inertia[:-1])

ax2.plot(k_range[1:], inertia_ratio, marker='o')
ax1.plot(k_range, inertia, marker='o')

#ax1.xlabel('Número de clusters')

#ax1.ylabel('Inercia')

#ax1.title('Método del Codo')

plt.show()
```

```python
np.array(inertia[:-1])
```

**Creación de clustering**

```python
kmeans = KMeans(n_clusters=4, random_state=42)

clientes['cluster'] = kmeans.fit_predict(scaled_df)

clientes.head()
```

```python
group_clust = clientes.groupby('cluster').agg({'Id_Cliente':'count'})
group_clust['Share']=group_clust['Id_Cliente']/len(clientes)
group_clust
```

# Conclusiones generales de la activiad

### ¿Cuales son los grupos más rentables? ¿Porque?

Este grupo esta representado por el numero 0, dado que cuando se obtiene la suma del total del valor de todas las ordenes que ha realizado este clientes (a través del producto de AVG_ORDER_VALUE * TOTAL_ORDERS) se obtiene de forma absoluta el grupo con el mayor income.

```python
clientes["total_ordenes"] = clientes["Ordenes"] * clientes["AVG_Orden_USD"]
```

```python
clientes.groupby('cluster')\
    .agg({'total_ordenes':'sum'})\
    .sort_values(
            'total_ordenes', 
            ascending=False
        )\
.head(1)
```

### ¿Cuales deberían de recibir algun tipo de incentivo?

Para el grupo más rentable, el cual a su vez es el que cuenta con la recencia más baja, me parece que la estrategía a seguir puede ser a través diseñar un programa VIP que premie la alta frecuencia, volumen y reciente actividad de compra. Beneficios como descuentos personalizados, acceso anticipado a nuevos lanzamientos y servicios exclusivos consolidarán la relación.

Desarrollar campañas que, basadas en el análisis de datos y segmentación por cuartiles, adapten el mensaje y canal para cada perfil. Así se asegura que el contenido resuene con sus expectativas y refuerce la imagen premium de la marca.

```python
responde = clientes.groupby('cluster')[['Ordenes','AVG_Orden_USD','Recencia', 'total_ordenes']].mean()
responde
```

### ¿Cual cluster son los menos rentables? ¿Porque?

Este grupo es el que tiene el menor ingreso absoluto, pero también es el que tiene la recencia más alta. Esto significa que estos clientes no han realizado compras recientemente, lo que indica una falta de interés.
