* En la parte de summary

  * H0 = Coeficiente = 0

**Coeficientes de variables numericas**

- Si el coeficiente es positivo, significa que a medida que la variable independiente aumenta, la variable dependiente también tiende a amentar.
- Si el coeficiente es negativo, significa que medida que la variable independiente aumenta, la variable dependiente tiende a disminuir

**Coeficientes de variables categóricas**
- Se interpretan en relación a la categoría de referencia (la que no tiene coeficiente)

- Si el coeficiente de una categoría es positivo, significa que esa categoría tiene un efecto positivo en la variable dependiente en comparación con la categoría de referencia.
- Si el coeficiente de una categria es negativo, significa que esa categoría tiene un efecto negativo en la variable depndiente en comparación con la categoría de referencia.

R-squared es una medida de la proporción de la varianza de la variable dependiente que es explicada por el modelo. Un valor más alto indica un mejor ajuste del modelo a los datos.
