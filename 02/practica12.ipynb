```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar dataset
data = pd.read_csv("./resources/modaexpress_ventas.csv")
```

```python
data.head(3)
```

```python
data.info()
```

```python
correlacion_matrix = data.corr(numeric_only = True)

plt.figure(figsize=(10, 8))
sns.heatmap(correlacion_matrix, annot=True, cmap='coolwarm')
plt.title('Maatriz de correlación entre variables')
plt.show()
```

```python
sns.scatterplot(data, x="Gasto Promociones", y="Ventas",hue="Region")

plt.show()
```
