#### **Punto 3 (35 puntos)**

Reliza una gráfico de scatterplot usando el precio y los caballos de fuerza e **interpreta**.

**Respuesta** Para el scatterplot, se logra apreciar que los datos siguen una tendencia creciente, es decir, mientra mayor sea el precio de estas los caballos de fuerza tenderan a incrementar en el vehiculo.

```
#Realiza el código correspondiente
import matplotlib.pyplot as plt

plt.scatter(df['Caballos de fuerza'], df['Precio'])

plt.title("Correlación entre caballos de fuerza y precio")
plt.xlabel("Caballos de fuerza")
plt.ylabel("Precio")

plt.show()
```
