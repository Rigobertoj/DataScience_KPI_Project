# La importancia de la identificación de la audiencia

Identificar quién recibirá nuestro mensaje es el primer paso para diseñar nuestras presentaciones, pues este punto determina el tono y el contenido que se desarrollará. En este sentido, la audiencia es el grupo de personas que escuchará nuestra presentación, y es importante tener en cuenta que no todas las audiencias son iguales. Por ejemplo, una presentación dirigida a un grupo de expertos en un tema específico será muy diferente a una presentación dirigida a un grupo de personas que no tienen conocimientos previos sobre el tema.

## Tipos de audiencia

### C-Level

- Intereses clave: Visión general de alto nivel, resultados financieros globales, retorno de inversión, impacto estratégico.

- Datos relevantes: Indicadores clave de desempeño (KPI), tendencias generales, comparativas rápidas.

F- ormato de presentación: Resumen ejecutivo con conclusiones puntuales. Gráficas sencillas y claras.

- Visual: Gráficos de barras o líneas que muestren la tendencia a lo largo del tiempo; textos destacados con números clave.

### Director

- Intereses clave: Integración de áreas, alineación con objetivos corporativos, oportunidades de mejora.

- Datos relevantes: Análisis comparativos entre departamentos, evolución de proyectos, metas intermedias.

- Formato de presentación: Documento con secciones claras y conclusiones que permitan ver relaciones entre diferentes áreas.

- Visual: Tablas resumidas y heatmaps para resaltar diferencias, gráficos comparativos (barras agrupadas).

### Gerente

- Intereses clave: Resultados detallados de su equipo, acciones concretas para la operación, control de procesos.

- Datos relevantes: Métricas específicas de desempeño operativo, planes de acción, proyecciones a corto plazo.

- Formato de presentación: Reportes más detallados, con explicaciones de hallazgos y pasos sugeridos.

- Visual: Gráficos de barras para desglose, tablas con KPIs diarios o semanales, diagramas de flujo.

## Preguntas

**¿Porqué es importante identificar el tipo de audiencia antes de realizar una presentación?** Esto tiene como objetivo desarrollar un mensaje más orientado a los intereses y necesidades de cada tipo de audiencia, de este modo el contenido será considerado de valor y podrá ser tomaro en cuenta en las decisiones de cada lider de la organización lo que reveleraria la importancia del uso de la información en las organizaciones y en su estructura pues sin esta cada una de las cabezas de la organización se encontraria perdida y no tendría visibilidad del los objetivos y metas de la empresa.
