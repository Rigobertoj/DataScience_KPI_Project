#### **Parte 4 - Cálculo de KPI e interpretaciones**

##### **Realiza los siguientes cálculos e iterpreta**

**Costo de Adquisición de Clientes**

Fórmula: $CAC = \frac{\text{Gastos en Marketing y Ventas}}{\text{Número de Clientes Nuevos}}$

**4.1 Calcula el CAC en cada unos de los años (2022 a 2024) e interpreta** (6 puntos)

Interpretación: Claramente dado que el año 2022 fue el año con la mayor cantidad de clientes nuevos, este mismo es el que menor costo de adquisición de clientes tiene.

```
costo.info()
```

```
#Code

costo["Fecha_mes"] = pd.to_datetime(costo["Fecha"].dt.strftime('%Y-%m-1'))

costo["Año"] = costo["Fecha"].dt.year

costo["Mes"] = costo["Fecha"].dt.month

df_costo = costo.groupby("Año")["Gasto_Marketing"].sum()

CAC = df_costo / df_primeros_clientes

CAC
```

**4.2 Calcula el CAC mensual desde 2022 a 2024 y realiza una grafica para representar el CAC a lo largo del tiempo.** (6 puntos)

```
#Code
df_costo_mensual = costo.groupby("Fecha_mes")["Gasto_Marketing"].sum()

CAC = df_costo_mensual / df_nuevos_clientes

CAC.plot()

plt.title("Tendencia del CAC a traves de los meses")
plt.xlabel("Fecha")
plt.ylabel("CAC")

ax = plt.gca()
ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))

plt.show()
```

**4.3 Da una recomendación a la empresa en relación al CAC** (5 puntos)

*Recomendación:* Desde mi punto de vista si hace falta una revisión a las estrategias de marketing pues mientras que la tendencia de la captación de nuevos clientes tiene una tendencia muy solida a converger a una punto constante, la tendencia del gasto en marketing se mantiene a la alza y no parece que haya variación significativa en los resultados de nuevos leads_

**Margen Bruto:**

Fórmula: $\text{Margen Bruto} = \frac{\text{Ingresos Totales} - \text{Costos Directos}}{\text{Ingresos Totales}}$

**4.4 Calcula el Margen Bruto mensual desde 2022 a 2024 y realiza una grafica para representar el Margen Bruto a lo largo del tiempo e interpreta.** (6 puntos)

Interpretación:______________

```
# Corrected Code

df_resultados_mensual = costo.copy()

result = ordenes.groupby("fecha_mes")["USD"].sum()

df_resultados_mensual["Ingresos"] = result.to_numpy()

df_resultados_mensual["Margen_bruto"] = (df_resultados_mensual["Ingresos"] - df_resultados_mensual["Costo_Directos"]) / df_resultados_mensual["Ingresos"]

# Plot results
sns.lineplot(data=df_resultados_mensual, x="fecha_mes", y="Margen_bruto")

plt.title("Tendencia del Margen Bruto a traves de los meses")
plt.xlabel("Fecha")
plt.ylabel("Margen Bruto")

plt.xticks(rotation=45)

plt.show()
```
