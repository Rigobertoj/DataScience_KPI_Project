# **Sesión 2: Entendiendo y clasificando indicadores**

## **Objetivo:**
Proporcionar a los estudiantes un entendimiento práctico de los diferentes tipos de indicadores clave de rendimiento (KPIs) y su clasificación, así como los conceptos básicos de unit economics. La sesión se enfocará en discutir conceptos clave y resolver preguntas del caso sin depender de programación compleja.

---

## **Estructura de la sesión:**

### **1. Introducción a los KPIs (20 minutos)**
- Breve explicación sobre qué son los KPIs y su importancia para las empresas.

  - Métrica que indica **DESEMPEÑO**. 
  - ¿Cómo es distinto a una simple métrica?
  - **¿Por qué es importante un KPI?**
    - **Monitoreo de progreso**
    - **Identificación de oportunidades de mejora**
    - **Facilita la toma de decisiones**

  - ¿Cómo deben ser?, ¿por qué? Recuerden **SMART** para que sea **EFECTIVO**
    - Specific
    - Measurable
    - Achievable
    - Relevant
    - Time-bound
- Clasificación básica de KPIs:
  - **Cuantitativos vs. Cualitativos:**
    - *Cuantitativos*: Medibles y expresados en números (e.g., Ingreso, Margen de Ganancia, Ventas Totales).
    - *Cualitativos*: Más subjetivos y basados en percepción (e.g., Satisfacción del Cliente, Reputación de Marca).
  - **Estratégicos, Operacionales y Funcionales:**
    - *Estratégicos*: Metas de alto nivel alineadas con la visión empresarial. (e.g., Market Share, Crecimiento MoM, Metas de ventas)
    - *Operacionales*: Enfocados en la eficiencia diaria. (e.g., Tiempos de respuesta al cliente, Tiempos de producción, Avg. ETA en un Uber)
    - *Funcionales*: Relacionados con departamentos específicos. (e.g. HR: Tasa de rotación de personal, employee engagement, Mkt: CAC, Growth)

### **2. Unit Economics y su importancia (30 minutos)**

- Nos vamos a concentrar en un **costo unitario**—¿cuánto gano o pierdo por cada venta o cada cliente?
- ¿Por qué son importantes los **unit economics**?
- **Conceptos clave:**

  - **Costo de Adquisición de Clientes (CAC):**
    - Fórmula: $CAC = \frac{\text{Gastos en Marketing y Ventas}}{\text{Número de Clientes Nuevos}}$
    - Indicador de cuánto cuesta adquirir un cliente.
    - Ejemplo: Si te cuesta $10,000MXN adquirir 200 clientes, su CAC es de $50MXN

  - **Valor de Vida del Cliente (LTV) a 24 meses:**
    - Fórmula: $LTV = \frac{\text{Ingresos Totales de los Primeros 24 Meses de la vida de los clientes}}{\text{Clientes con 24 meses de tenure}}$
    - Mide el valor que un cliente genera durante sus primeros 24 meses con la empresa.
    - Ejemplo: Si en 24 meses, 100 clientes te generan $10,000MXN su LTV24 es de $100MXN pero **NO CONSIDERA OTROS COSTOS**

  - **Margen Bruto:**
    - Fórmula: $\text{Margen Bruto} = \frac{\text{Ingresos Totales} - \text{Costos Directos}}{\text{Ingresos Totales}}$
    - Indica la rentabilidad después de cubrir los costos directos.
    - Ejemplo: Si vendes $10,000MXN y tus costos de producción son de $6,000MXN tu margen bruto es de 40%

### **3. Clasificación práctica de indicadores (30 minutos)**
    - Cuantitativos
    - Cualitativos
    ----------------------
    - Estratégicos
    - Operacionales
    - Funcional
- **Ejemplos por categoría:**
    - Ingreso                               -> Cuantitivo / Estrategico
    - Margen de Ganancia                    -> Cuantitivo / Estrategico
    - Tasa de Productos Defectuosos         -> Cuantitivo / Operacional
    - LTV                                   -> Cuantitivo / Estrategico
    - CTR (Click Through Rate)              -> Cuantitivo / Funcional
    - Retorno sobre la Inversión (ROI)      -> Cuantitivo / Estrategico
    - Costo por Lead (Costo Marketing / #clientes generados)                      -> Cuantitivo / Funcional
    - On-time Delivery Rate                 -> Cuantitivo / Operacional
    - Tasa de Conversión                    -> Cuantitivo / Funcional
    - Participación de Mercado              -> Cuantitivo / Estrategico
    - Employee Engagement                   -> Cualitativo / Funcional
    - Tiempo de Entrega de Pedidos          -> Cuantitivo / Operacional
