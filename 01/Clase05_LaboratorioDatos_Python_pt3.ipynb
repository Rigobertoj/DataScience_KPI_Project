### **Caso práctico:**

Se tiene una base de datos de los clientes de una empresa con su comportamiento de compra en los meses de Enero a Abril del 2020 y 2021, con la siguiente información:

- id: Número único de registro del cliente.

- ciudad: Ciudad donde se encuentra la tienda.

- edad: Edad actual.

- genero: Género (Female / Male).

- mes: Mes en el que se realizó la orden.

- ordenes: Número de órdenes generadas en el mes.

- paid: Variable que indica si las órdenes fueron pagadas o no (1 Si, 0 No)

- usd: Dólares del total de órdenes generadas.

### 1.1 Asignar los grupos de edad. 

**HINT**: Usar función `pd.cut()`

```python

bins = [0, 18, 25, 41, 57, 67, 100] 
labels = ["Menor de edad", "Gen Z", "Millennials", "Gen X", "Boomers", "Otra generación"]

ventas["Generacion"] = pd.cut(ventas["edad"], bins=bins, labels=labels, right=True)

ventas.head(5)
```

```python
ventas['ordenes'].unique()
```

### 1.2 Calcular para los meses de 2021: 

**HINT**: Es necesario primero convertir la columna 'mes' a datetime. 

- Total de órdenes mensuales

- Promedio mensual de USD (Total de USD por mes / Total de órdenes)

- Total de USD

```python
ventas["mes"] = pd.to_datetime(ventas["mes"], format="%b %y")

ventas.head(5)
```

- Total de órdenes mensuales

```python
filtro_mes = pd.to_datetime("2021-01-01")

ventas_mes = ventas[ventas["mes"] >= filtro_mes].groupby('mes')['ordenes'].sum()

ventas_mes
```

- Promedio mensual de USD por orden (Suma de USD / Total de órdenes)

```python
ventas_ordenes_promedo = ventas[ventas["mes"] >= filtro_mes].groupby("mes").agg({"usd" : "sum", "ordenes" :"sum"})
ventas_ordenes_promedo["result"] = ventas_ordenes_promedo["usd"]/ventas_ordenes_promedo["ordenes"]
ventas_ordenes_promedo["result"]
```

- Total de USD por mes

```python
ventas_mes_usd = ventas[ventas['mes'] >= filtro_mes].groupby("mes")["usd"].sum()

ventas_mes_usd.head()
```

### 1.3 Cálculo después de 2021:

- ¿Quién es el grupo de edad con más ventas totales después del 2021?

- ¿Cuántas órdenes hay sin pagar del mes de Febrero de 2021?

```python
fecha = pd.to_datetime("2021-01-01")

grupos_edad_filtro = ventas[ventas["mes"] >= fecha]

grupos_edad = grupos_edad_filtro.groupby("Generacion")["usd"].sum().idxmax()
grupos_edad
```

- ¿Cuántas órdenes hay sin pagar del mes de Febrero de 2021?

```python
fechas_filtro = pd.to_datetime("2021-02-01")

ventas_2021_02 = ventas[ventas['mes'] == fechas_filtro]

total_ordenes_sin_pago = ventas_2021_02[ventas_2021_02["pago"] == 0]['id'].count()

total_ordenes_sin_pago
```

- Agrega un impuesto del 16% a aquellas órdenes que sean compradas en Guadalajara

**HINT**: Usa  `apply` y `lambda` sobre las columnas de `usd` y `ciudad`.

Por ejemplo: 

`df[['columna_a','columna_b']].apply(lambda x: 1 if x['columna_a'] > x['columna_b'] else 0, axis = 1)`

`axis = 1` se usa cuando aplicas la función sobre más de una columna

```python
ventas.head()
```

```python
def impuesto_guadalajara(price):
    return price * 1.16

fecha = pd.to_datetime("2021-01-01")

filtro_edad = ventas[ventas["mes"] >= fecha]

filtro_edad["usd_con_impuesto"] = filtro_edad[["ciudad", "usd"]].apply(
    lambda x: x["usd"] * 1.16 if x["ciudad"] == "Guadalajara" else x["usd"], axis=1
)

filtro_edad.head()
```
