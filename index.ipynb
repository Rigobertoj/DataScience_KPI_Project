```python
def calculo_cac(marketing_expense : float, num_customer : float) -> float:
    """
    summary: Calculo de adquisión por cliente (CAC)

    Args:
        - marketing_expense (float): Costos de adqusisión del cliente
        - num_customer (float): numero de clientes

    Returns:
        float: CAC
    """
    
    CAC = marketing_expense/num_customer
    
    return CAC

for i in range (10):
    print(i)
```

```python
import matplotlib.pyplot as plt

# Valores posibles de la suma (2 a 12)
x = range(2, 13)

# Probabilidades correspondientes
pmf = [1/36, 2/36, 3/36, 4/36, 5/36, 6/36, 5/36, 4/36, 3/36, 2/36, 1/36]

plt.bar(x, pmf)
plt.xlabel('Suma de los dados')
plt.ylabel('Probabilidad')
plt.xticks(x)
plt.title('Función de masa de probabilidad para la suma de dos dados')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Valores posibles de la suma (2 a 12)
x = range(2, 13)

# Probabilidades correspondientes (PMF)
pmf = [1/36, 2/36, 3/36, 4/36, 5/36, 6/36, 5/36, 4/36, 3/36, 2/36, 1/36]

# Calculamos la distribución acumulada (CDF)
cdf = []
suma_acum = 0
for p in pmf:
    suma_acum += p
    cdf.append(suma_acum)

print(cdf)
# Graficamos la CDF
#plt.plot(x, cdf, marker='o')
#plt.xlabel('Suma de los dados')
#plt.ylabel('Probabilidad Acumulada')
#plt.title('Distribución acumulada (CDF) de la suma de dos dados')
#plt.xticks(x)
#plt.show()
```
